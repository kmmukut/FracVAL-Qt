# FracVAL: generate and visualize on demand
This notebook uses the Python API backed by the Fortran PCA/CCA engine. Run `make python-ext` first.


```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if ROOT.name == 'plot': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'python'))
from fracval import FracVALConfig, generate, plot_3d
```

## Monodisperse aggregate


```python
mono = generate(FracVALConfig(n=100, df=1.79, kf=1.40, rp_g=15.0, rp_gstd=1.0, seed=12345))
mono.metadata()
```

```python
plot_3d(mono, mode='spheres', sphere_resolution=9).show()
```

## Polydisperse aggregate


```python
poly = generate(FracVALConfig(n=100, df=1.68, kf=0.98, rp_g=15.0, rp_gstd=2.0, seed=67890))
plot_3d(poly, mode='centers').show()
```
